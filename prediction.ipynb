```python
from keras.datasets import imdb
from keras.preprocessing import sequence
from keras.models import load_model
```

```python
word_index = imdb.get_word_index()
reverse_word_index = {value : key for key, value in word_index.items()}
```

```python
model = load_model("simpleRNN.h5")
model.summary()
```

```python
for i, w in enumerate(model.get_weights()):
    print(f"weight {i}: shape {w.shape}")
```

```python
def decoded_review(encoded_review):
    for i in encoded_review:
        word = reverse_word_index.get(i-3)
        # i-3 due to offset this code is in the site
        if word == None:
            print("__________", end = ' ')
            continue
        print(word, end = " ")

def preprocess_text(text):
    words = text.lower().split()

    encoded_review = [1] 
    for word in words:
        encoded = word_index.get(word, 2) + 3
     
        if encoded >= 10000:
            encoded = 2 

            
        encoded_review.append(encoded)

    padded_review = sequence.pad_sequences(
        [encoded_review],
        maxlen=500
    )

    return padded_review
```

```python
def prediction(review):
    preprocessed_input = preprocess_text(review)
    prediction = model.predict(preprocessed_input)
    if prediction[0][0] > 0.5:
        result = "Positive";
    else:
        result = "Negative"
    return prediction[0][0], result
```

```python
sample_review = "Interstellar is a beautiful and emotionally powerful movie. The story combines space exploration, science, family, and human survival in a very engaging way. The visuals are stunning, and the music creates an incredible atmosphere. The emotional relationship between Cooper and his daughter makes the movie much more than just a science-fiction film. Some parts are slow and the scientific concepts can be difficult to understand, but the overall experience is unforgettable. It is a thoughtful movie that stays in your mind long after it ends."

score, result = prediction(sample_review)
print(f"Sentiment: {result} | Positive score: {score:.4f}")
```
