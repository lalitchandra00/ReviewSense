```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
word_index = imdb.get_word_index()
reverse_word_index = {value : key for key, value in word_index.items()}
```

```python
model = load_model("simpleRNN.h5")
model.summary()
```

```python
def decoded_reviwe(encoded_review):
    for i in encoded_review:
        word = reverse_word_index.get(i-3)
        # i-3 due to offset this code is in the site
        if word == None:
            print("__________", end = ' ')
            continue
        print(word, end = " ")

def preprocess_text(text):
    words = text.lower().split()

    encoded_review = [
        word_index.get(word, 2) + 3
        for word in words
    ]

    padded_review = sequence.pad_sequences(
        [encoded_review],
        maxlen=500
    )

    return padded_review
```
